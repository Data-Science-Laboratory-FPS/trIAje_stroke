```python
key_col = 'DEMANDAPK'
id_columns = ['IDTARJETASANITARIA', 'DNI', 'NOMBRE_APELLIDOS']
```

```python
# Loading exported table
master_table = pq.read_table(os.path.join(settings.source_tables_path, settings.master_table_path)).to_pandas()
master_table.shape
```

```python
master_table.head()
```

```python
dp.summarize_numbers(master_table, key_col, id_columns)
```

# EDATOSCASO-LLAMADAS 5 DEMAND TYPES

```python
master_table["IDTIPODEMANDA1"].value_counts(ascending=True)
```

```python
dp.summarize_numbers(master_table_4demands, key_col, id_columns)
```

```python
# Data Export
master_table_4demands.to_parquet(os.path.join(settings.source_tables_path, settings.master_table_4demands_path), index=False)
```
